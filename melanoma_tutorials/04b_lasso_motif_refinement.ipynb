# CISM Tutorial 04b: LASSO Motif Refinement

This notebook displays the outputs produced by `04b_lasso_motif_refinement.py`. Run the script first when the output directory does not exist or when you want to refresh the analysis.

```python
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

TUTORIAL_DIR = Path.cwd()
if TUTORIAL_DIR.name != "melanoma_tutorials":
    TUTORIAL_DIR = Path("melanoma_tutorials")

OUTPUT_DIR = TUTORIAL_DIR / "melanoma_tutorials" / "runtime" / "lasso_motif_refinement"
IMAGE_DIR = OUTPUT_DIR / "images"

print(f"04b output directory: {OUTPUT_DIR.resolve()}")
```

## Original vs Pruned vs Refined Panels

```python
comparison_metrics = pd.read_csv(OUTPUT_DIR / "panel_comparison_metrics.csv")
display(comparison_metrics)

image_path = IMAGE_DIR / "panel_comparison_metrics.png"
if image_path.exists():
    display(Image(filename=image_path))
```

## LASSO Feature Importance

```python
lasso_importance = pd.read_csv(OUTPUT_DIR / "lasso_feature_importance.csv")
display(lasso_importance.head(30))

image_path = IMAGE_DIR / "lasso_feature_importance.png"
if image_path.exists():
    display(Image(filename=image_path))
```

## Removed Top Motifs

```python
removed_top = pd.read_csv(OUTPUT_DIR / "lasso_removed_top_motifs.csv")
display(removed_top)

image_path = IMAGE_DIR / "lasso_removed_top_motifs_bar.png"
if image_path.exists():
    display(Image(filename=image_path))
```

## Lower-Ranked Candidate Trials

```python
candidate_trials = pd.read_csv(OUTPUT_DIR / "replacement_trials.csv")
display(candidate_trials)

image_path = IMAGE_DIR / "replacement_decisions.png"
if image_path.exists():
    display(Image(filename=image_path))
```

## Final Motif Panel for Deciphering

```python
final_panel = pd.read_csv(OUTPUT_DIR / "final_refined_motif_panel.csv")
display(final_panel)

image_path = IMAGE_DIR / "final_refined_motif_panel_bar.png"
if image_path.exists():
    display(Image(filename=image_path))
```

## Asymmetric Dominant/Minority Stringency

```python
asymmetric_tuning_path = OUTPUT_DIR / "asymmetric_stringency_tuning.csv"
asymmetric_selected_path = OUTPUT_DIR / "asymmetric_stringency_selected_motifs.csv"
asymmetric_metrics_path = OUTPUT_DIR / "asymmetric_stringency_metrics.csv"

if asymmetric_tuning_path.exists():
    asymmetric_tuning = pd.read_csv(asymmetric_tuning_path)
    asymmetric_selected = pd.read_csv(asymmetric_selected_path)
    asymmetric_metrics = pd.read_csv(asymmetric_metrics_path)

    display(asymmetric_tuning.head(20))
    display(asymmetric_metrics)
    display(asymmetric_selected)

    for name in [
        "asymmetric_stringency_tuning.png",
        "asymmetric_stringency_selected_motifs_bar.png",
        "asymmetric_stringency_roc_curve.png",
    ]:
        image_path = IMAGE_DIR / name
        if image_path.exists():
            display(Image(filename=image_path))
else:
    print("Run 04b_lasso_motif_refinement.py without --skip-asymmetric-stringency to generate this experiment.")
```

## Refined Panel Validation

```python
refined_metrics = pd.read_csv(OUTPUT_DIR / "refined_panel_metrics.csv")
refined_results = pd.read_csv(OUTPUT_DIR / "refined_panel_loocv_results.csv")

display(refined_metrics)
display(refined_results.head())

for name in ["refined_panel_confusion_matrix.png", "refined_panel_roc_curve.png"]:
    image_path = IMAGE_DIR / name
    if image_path.exists():
        display(Image(filename=image_path))
```
